```python
import csv
```

```python
filename = "R1-1711982_BG1.csv"

rows = []
total_data = []
rowOffset = [0]
rowWeight = []
columnIndexMap = []


with open(filename, 'r', encoding='utf-8-sig') as csvfile:
    # creating a csv reader object
    csvreader = csv.reader(csvfile)

    # extracting each data row one by one
    for row in csvreader:
        rows.append(row)

    # get total number of rows
    print("Total no. of rows: %d" % (csvreader.line_num))
```

```python
for row in rows:
    rowWeightNum = 0
    for i in range (len(row)):
        if (int(row [i]) >= 0):
            total_data.append (int(row [i]))
            columnIndexMap.append(i)
            rowWeightNum += 1
    rowWeight.append (rowWeightNum)
            
```
